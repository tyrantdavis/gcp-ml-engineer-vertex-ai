## Separation of Concerns (Trainer vs Pipeline vs Deployment)

- Why production ML separates:

  Production systems deliberately isolate responsibilities to reduce
  coupling, improve reliability, and enable independent evolution.

- Trainer logic  

  Encapsulates model definition, data loading, and training behavior.

- Pipeline orchestration  

  Manages execution order, parameterization, retries, and reproducibility.

- Deployment configuration  

  Defines how models are served, versioned, updated, and monitored.

## Training Pipelines as First-Class Artifacts

- What makes a pipeline “production-grade”  

  Pipelines are infrastructure, not wrappers around notebooks.

- Idempotency  

  Pipelines must be safe to re-run without corrupting system state.

- Parameterization  

  Configuration should change behavior without modifying code.

- Reusability  

  Pipelines should support multiple runs, datasets, and environments.
